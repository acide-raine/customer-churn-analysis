```python
from scr.exp import explore, data_visualizator
from scr import confiq
```

Общая инфа по датасету

```python
confiq.TRAIN_DF.info()
```

Столбец TotalSpent проблемный. Значения записаны в str хотя фактически там float. Ниже функция конвертации

```python
DF = explore.convert_totalspent_to_num(confiq.TRAIN_DF, "TotalSpent")
```

Разделяем данные по категориям для более комфортные работы

```python
category, numbers, binary = explore.type_collect(DF, confiq.TARGET_COLUM)
```

Проходим по каждой категории признаков, собираем стат инфу. Выводим собранную стат данные

Категориальные признаки

```python
category_stat_info = explore.categorical_futures_analiz(DF, category)

for col,data in category_stat_info.items():
    print(f"Колонка: {col}")
    for k, v in data.items():
        print(f"Значение {k}: {v}")
    print()
```

Числовые признаки:

```python
numeric_stat_info = explore.numeric_futures_analiz(DF, numbers)

for col,data in numeric_stat_info.items():
    print(f"Колонка: {col}")
    for k, v in data.items():
        print(f"Значение {k}: {v}")
    print()
```

Бинарные числовые и строковые признаки

```python
binary_stat_info = explore.binary_futures_analiz(DF, binary)

for col,data in binary_stat_info.items():
    print(f"Колонка: {col}")
    for k, v in data.items():
        print(f"Значение {k}: {v}")
    print()
```

Визуализируем полученные данные

Распределения категориальных признаков:

```python
for i in range(len(category)):
    data_visualizator.distributions(DF, category[i])
```

Распределения числовых признаков:

```python
for i in range(len(numbers)):
    data_visualizator.distributions(DF, numbers[i])
```

Распределение бинарных признаков:

```python
for i in range(len(binary)):
    data_visualizator.distributions(DF, binary[i])
```

Собираем матрицы корреляций

```python
pearson_n_cor, spearman_s_cor = explore.num_correlations(DF, numbers)
categorical_correlations = explore.cramer_v_matrix(DF, category)
binary_correlations = explore.str_binary_correlation(DF, binary)

target_correlation = explore.target_corr(DF, confiq.TARGET_COLUM, numbers, category, binary)
```

Корреляция числовых признаков

```python
data_visualizator.num_cor(pearson_n_cor, spearman_s_cor)
```

Корреляция категориальных признаков

```python
data_visualizator.cat_cor(categorical_correlations, binary_correlations)
```

Кореляция с таргетом

```python
data_visualizator.target_cor(target_correlation)
```
